```python
from collections import Counter

X = int(input())
shoes = list(map(int, input().split()))

N = int(input())

shoe_count = Counter(shoes)
total = 0

for _ in range(N):
    size, price = map(int, input().split())

    if shoe_count[size] > 0:
        total += price
        shoe_count[size] -= 1

print(total)
```
